```python
import os
import pandas as pd

# Set the project folder
folder = r"C:\Users\DELL\Downloads\Supply_Chain_Inventory_Analytics_Starter\Supply_Chain_Inventory_Analytics"

# Load all five datasets
products = pd.read_csv(os.path.join(folder, "products.csv"))
suppliers = pd.read_csv(os.path.join(folder, "suppliers.csv"))
inventory = pd.read_csv(os.path.join(folder, "inventory_snapshot.csv"))
sales = pd.read_csv(os.path.join(folder, "sales_orders.csv"))
purchase = pd.read_csv(os.path.join(folder, "purchase_orders.csv"))

print("All datasets loaded successfully!")
```

```python
# Convert date columns to datetime format
sales["order_date"] = pd.to_datetime(sales["order_date"])

inventory["snapshot_date"] = pd.to_datetime(
    inventory["snapshot_date"]
)

purchase["order_date"] = pd.to_datetime(purchase["order_date"])
purchase["expected_delivery_date"] = pd.to_datetime(
    purchase["expected_delivery_date"]
)
purchase["actual_delivery_date"] = pd.to_datetime(
    purchase["actual_delivery_date"]
)

# Check inventory calculations
inventory["calculated_available_stock"] = (
    inventory["stock_on_hand"] - inventory["reserved_stock"]
)

print("Inventory calculation mismatches:",
      (inventory["available_stock"] !=
       inventory["calculated_available_stock"]).sum())

# Check sales revenue calculations
sales["calculated_revenue"] = (
    sales["quantity_sold"] * sales["unit_price"]
).round(2)

print("Sales revenue mismatches:",
      (sales["sales_revenue"].round(2) !=
       sales["calculated_revenue"]).sum())

# Check date ranges
print("\nSales date range:")
print(sales["order_date"].min(), "to", sales["order_date"].max())

print("\nPurchase order date range:")
print(purchase["order_date"].min(), "to", purchase["order_date"].max())

print("\nData validation completed!")
```

```python
# 1. Total sales revenue
total_revenue = sales["sales_revenue"].sum()

# 2. Total units sold
total_units_sold = sales["quantity_sold"].sum()

# 3. Total inventory value
inventory["inventory_value"] = (
    inventory["stock_on_hand"] * inventory["unit_cost"]
)
total_inventory_value = inventory["inventory_value"].sum()

# 4. Products below reorder level
low_stock = inventory[
    inventory["available_stock"] < inventory["reorder_level"]
]
low_stock_count = low_stock["product_id"].nunique()

# 5. Supplier delivery performance
on_time_orders = (purchase["delivery_delay_days"] == 0).sum()
total_purchase_orders = len(purchase)

on_time_delivery_rate = (
    on_time_orders / total_purchase_orders
) * 100

# Display KPIs
print("===== SUPPLY CHAIN & INVENTORY KPIs =====")
print(f"Total Sales Revenue: ₹{total_revenue:,.2f}")
print(f"Total Units Sold: {total_units_sold:,}")
print(f"Total Inventory Value: ₹{total_inventory_value:,.2f}")
print(f"Products Below Reorder Level: {low_stock_count}")
print(f"On-Time Delivery Rate: {on_time_delivery_rate:.2f}%")
```

```python
# Display products below reorder level
low_stock_products = low_stock.merge(
    products[["product_id", "product_name", "category"]],
    on="product_id",
    how="left"
)

low_stock_products = low_stock_products[
    [
        "product_id",
        "product_name",
        "category",
        "available_stock",
        "reorder_level"
    ]
]

low_stock_products = low_stock_products.sort_values(
    "available_stock"
)

display(low_stock_products)
```

```python
# Calculate supplier delivery performance
supplier_performance = purchase.groupby("supplier_id").agg(
    total_orders=("po_id", "count"),
    on_time_orders=("delivery_delay_days",
                    lambda x: (x == 0).sum()),
    avg_delay_days=("delivery_delay_days", "mean"),
    avg_lead_time=("lead_time_days", "mean")
).reset_index()

# Calculate on-time delivery percentage
supplier_performance["on_time_delivery_rate"] = (
    supplier_performance["on_time_orders"] /
    supplier_performance["total_orders"] * 100
).round(2)

# Add supplier names
supplier_performance = supplier_performance.merge(
    suppliers[["supplier_id", "supplier_name", "region"]],
    on="supplier_id",
    how="left"
)

# Sort by average delay, highest first
supplier_performance = supplier_performance.sort_values(
    "avg_delay_days", ascending=False
)

display(supplier_performance)
```

```python
monthly_sales = (
    sales.set_index("order_date")
    .resample("MS")["sales_revenue"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales["order_date"],
    monthly_sales["sales_revenue"],
    marker="o"
)

plt.xlabel("Month")
plt.ylabel("Sales Revenue (₹)")
plt.title("Monthly Sales Revenue Trend")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Sales performance by category
category_sales = sales.merge(
    products[["product_id", "category"]],
    on="product_id",
    how="left"
)

category_sales_summary = category_sales.groupby("category").agg(
    total_revenue=("sales_revenue", "sum"),
    total_units_sold=("quantity_sold", "sum"),
    total_orders=("sales_order_id", "count")
).reset_index()

category_sales_summary = category_sales_summary.sort_values(
    "total_revenue", ascending=False
)

display(category_sales_summary)
```

```python
# Inventory value by category
category_inventory = inventory.merge(
    products[["product_id", "category"]],
    on="product_id",
    how="left"
)

category_inventory_summary = category_inventory.groupby("category").agg(
    total_stock=("stock_on_hand", "sum"),
    total_inventory_value=("inventory_value", "sum"),
    total_products=("product_id", "nunique")
).reset_index()

category_inventory_summary = category_inventory_summary.sort_values(
    "total_inventory_value", ascending=False
)

display(category_inventory_summary)
```

```python
import sqlite3

# Create a database file
conn = sqlite3.connect("supply_chain.db")

# Export DataFrames as SQL tables
products.to_sql("products", conn, if_exists="replace", index=False)
suppliers.to_sql("suppliers", conn, if_exists="replace", index=False)
inventory.to_sql("inventory", conn, if_exists="replace", index=False)
sales.to_sql("sales", conn, if_exists="replace", index=False)
purchase.to_sql("purchase", conn, if_exists="replace", index=False)

print("All 5 tables created successfully!")
```

```python
query = """
SELECT
    p.category,
    SUM(s.sales_revenue) AS total_revenue,
    SUM(s.quantity_sold) AS total_units_sold,
    COUNT(s.sales_order_id) AS total_orders
FROM sales s
JOIN products p
    ON s.product_id = p.product_id
GROUP BY p.category
ORDER BY total_revenue DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    p.product_id,
    p.product_name,
    p.category,
    i.available_stock,
    i.reorder_level,
    (i.reorder_level - i.available_stock) AS stock_shortage
FROM inventory i
JOIN products p
    ON i.product_id = p.product_id
WHERE i.available_stock < i.reorder_level
ORDER BY stock_shortage DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    strftime('%Y-%m', order_date) AS sales_month,
    ROUND(SUM(sales_revenue), 2) AS total_revenue,
    SUM(quantity_sold) AS total_units_sold
FROM sales
GROUP BY strftime('%Y-%m', order_date)
ORDER BY sales_month;
"""

monthly_sql = pd.read_sql_query(query, conn)
display(monthly_sql)
```

```python
query = """
SELECT
    s.supplier_id,
    s.supplier_name,
    s.region,
    COUNT(p.po_id) AS total_orders,

    SUM(
        CASE
            WHEN p.delivery_delay_days = 0 THEN 1
            ELSE 0
        END
    ) AS on_time_orders,

    ROUND(AVG(p.delivery_delay_days), 2) AS avg_delay_days,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN p.delivery_delay_days = 0 THEN 1
                ELSE 0
            END
        ) / COUNT(p.po_id),
        2
    ) AS on_time_delivery_rate

FROM purchase p
JOIN suppliers s
    ON p.supplier_id = s.supplier_id

GROUP BY
    s.supplier_id,
    s.supplier_name,
    s.region

ORDER BY avg_delay_days DESC;
"""

supplier_sql = pd.read_sql_query(query, conn)
display(supplier_sql)
```
